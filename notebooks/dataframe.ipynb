```python
import os
import shutil

# make sure the 'bw' folder exists (create if not)
os.makedirs("bw", exist_ok=True)

bw = 0
for idx, row in df.iterrows():
    if 10 < row['ecc_lnB'] < 30:
        bw += 1
        sample_name = row['sample']

        # build full source and destination paths
        src = os.path.join("data", f"eccentric_{sample_name}.npy")
        dst = os.path.join("bw", f"eccentric_{sample_name}.npy")

        # move the file
        if os.path.exists(src):
            shutil.move(src, dst)
        else:
            print(f"File not found: {src}")

print("Count:", bw)
```

```python
import os
import shutil

# make sure the 'bw' folder exists (create if not)
os.makedirs("above", exist_ok=True)

above = 0
for idx, row in df.iterrows():
    if row['ecc_lnB'] >= 30:
        above += 1
        sample_name = row['sample']

        # build full source and destination paths
        src = os.path.join("data", f"eccentric_{sample_name}.npy")
        dst = os.path.join("above", f"eccentric_{sample_name}.npy")

        # move the file
        if os.path.exists(src):
            shutil.move(src, dst)
        else:
            print(f"File not found: {src}")

print("Count:", above)
```
