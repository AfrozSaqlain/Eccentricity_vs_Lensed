```python
import matplotlib.pyplot as plt
from multiprocessing import Pool
from gwtorch.modules.gw_utils import generate_prior, generate_waveform
```

```python
samples = generate_prior(1000)
```

```python
ecc_durations = []
lensed_durations = []
unlensed_durations = []
```

```python
with Pool(processes=14) as p:
    results = p.starmap(generate_waveform, [(i, samples[i]) for i in range(len(samples))])
```

```python
for result in results:
    ecc_durations.append(result[0])
    lensed_durations.append(result[1])
    unlensed_durations.append(result[2])
```

```python
plt.hist(ecc_durations, bins=30, alpha=0.5, label='Eccentric')
plt.hist(lensed_durations, bins=30, alpha=0.5, label='Lensed')
plt.hist(unlensed_durations, bins=30, alpha=0.5, label='Unlensed')
plt.legend()
plt.show()
```
