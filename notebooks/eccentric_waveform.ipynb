```python
from matplotlib import rc
rc('text', usetex=True)
rc('font', family='serif')
rc('font', serif='times')
#rc('font', weight='bolder')
rc('mathtext', default='sf')
rc("lines", markeredgewidth=1)
rc("lines", linewidth=1)
rc('axes', labelsize=20) #24
rc("axes", linewidth=0.5) #2)
rc('xtick', labelsize=10)
rc('ytick', labelsize=10)
rc('legend', fontsize=10) #16
rc('ytick', right='True',direction= 'in')
rc('xtick', top='True',direction= 'in')
rc('xtick.major', pad=15) #8)
rc('ytick.major', pad=15) #8)
rc('xtick.major', size=12) #8)
rc('ytick.major', size=12) #8)
rc('xtick.minor', size=7) #8)
rc('ytick.minor', size=7) #8)

def set_tick_sizes(ax, major, minor):
    for l in ax.get_xticklines() + ax.get_yticklines():
        l.set_markersize(major)
    for tick in ax.xaxis.get_minor_ticks() + ax.yaxis.get_minor_ticks():
        tick.tick1line.set_markersize(minor)
        tick.tick2line.set_markersize(minor)
    ax.xaxis.LABELPAD=10.
    ax.xaxis.OFFSETTEXTPAD=10.
```

```python
%matplotlib inline

import gwmat
import numpy as np
from gwmat import point_lens
import matplotlib.pyplot as plt
from pycbc.types import FrequencySeries, TimeSeries
from pycbc.waveform import get_fd_waveform, get_td_waveform
import matplotlib.pyplot as plt

mass1, mass2 = 50, 50
```

```python
ecc = [0.05, 0.1, 0.2]

hp_ecc = {}

for i, e in enumerate(ecc):
    hp, hc = get_td_waveform(approximant = 'teobresums',
                            mass1 = mass1, mass2 = mass2,
                            eccentricity = e, lambda1 = 0, lambda2 = 0,
                            spin1z = 0.0, spin2z = 0.0,
                            f_lower = 10, delta_t = 1.0/4096)

    hp_ecc[i] = FrequencySeries(hp.numpy(), delta_f=hp.delta_f)
```

## We will plot the squared norm of the Fourier components of the lensed and unlesned signal for different values of lens mass

```python
color = ["orange", "green", "red"]
```

```python
num_curves = len(ecc)

linewidths = np.linspace(3, 1, num_curves)

plt.figure(figsize=(10, 8))
plt.title(rf'$M_1 = M_2 = {mass1} M_\odot$')

for i in range(num_curves):
    plt.loglog(hp_ecc[i].sample_frequencies, np.sqrt(hp_ecc[i].squared_norm()),
               label=fr'e = {ecc[i]}',
               color=color[i],
               linewidth=linewidths[i])

plt.xlabel(r"Frequency (Hz)")
plt.ylabel(r'$|\tilde{h}(f)|^2$ [1/Hz]')
plt.xlim(10, 1024)
plt.legend()

ax = plt.gca()
set_tick_sizes(ax, major=12, minor=7)
plt.tight_layout()
plt.show()
```

```python
mass1, mass2 = 50, 50

hp, hc = get_td_waveform(approximant = 'IMRPhenomPv2',
                            mass1 = mass1, mass2 = mass2,
                            f_lower = 10, delta_t = 1.0/1024)

hp_fd = hp.to_frequencyseries()
```

```python
freq = hp_fd.sample_frequencies
Mlz = [10,1e2,1e3]
y = 0.8

hp_lensed = {}
t_delay = {}

for i in range(len(Mlz)):
    Ffs = [gwmat.cythonized_point_lens.Ff_effective(f, ml=Mlz[i], y=y) for f in freq]
    t_delay[i] = point_lens.time_delay(ml=Mlz[i], y=y)

    hp_lensed[i] = FrequencySeries(np.conj(Ffs) * hp_fd.numpy(), delta_f=hp_fd.delta_f, epoch=hp_fd.epoch).to_timeseries().crop(10.5, 4.9)#.cyclic_time_shift(-(t_delay[i] - 0.1))

hp = hp.to_timeseries().crop(10.5, 4.9)
```

```python
num_curves = len(Mlz)
linewidths = np.linspace(3, 1, num_curves)
alpha = np.linspace(0.5, 0.9, num_curves)

plt.figure(figsize=(10, 8))

plt.plot(hp.sample_times, hp, label='Unlensed', color='blue', linewidth=4)

for i in range(num_curves):
    plt.plot(hp_lensed[i].sample_times, hp_lensed[i],
               label=fr'Lensed $M_\odot$ = {Mlz[i]}',
               color=color[i],
               linewidth=linewidths[i],
               alpha=alpha[i])


plt.xlabel(r"Time (s)")
plt.ylabel(r'$h(t)$')
plt.legend(loc='upper left')

ax = plt.gca()
ax.text(-0.4, -1.5e-18, rf'$M_1 = M_2 = {mass1} M_\odot$, $y = 0.8$', fontsize=14)

set_tick_sizes(ax, major=12, minor=7)

plt.tight_layout()
plt.savefig('TD_lensed_waveform.pdf', bbox_inches='tight', dpi=300)
plt.show()
```
